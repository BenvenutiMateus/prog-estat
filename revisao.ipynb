## Passo 1 – Criar o dicionário principal
As chaves devem ser nomes de distribuições: `"Normal"`, `"Exponencial"`, `"Uniforme"`  
Os valores devem ser dicionários contendo:  
- `"params"` → parâmetros da distribuição  
- `"dados"` → inicialmente `None`  
- `"estatisticas"` → inicialmente `None`  

Exemplo de estrutura:

    distribuicoes = {
        "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
        "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
        "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
    }

```python
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

Criação do dicionário

## Passo 2 – Importar bibliotecas
Importe as bibliotecas:  
- `numpy` para simulação  
- `pandas` para organização dos resultados  
- `matplotlib` para visualização  

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

Importação das bibliotecas

## Passo 3 – Gerar os dados
Para cada distribuição, gere uma amostra de tamanho **1000** usando `numpy.random` e salve em `"dados"`.  

```python
np.random.seed(67)
distribuicoes['Normal']['dados'] = np.random.normal(**distribuicoes['Normal']['params'], size = 1000)
distribuicoes['Exponencial']['dados'] = np.random.exponential(**distribuicoes['Exponencial']['params'], size = 1000)
distribuicoes['Uniforme']['dados'] = np.random.uniform(**distribuicoes['Uniforme']['params'], size = 1000)
```

Geração dos dados a partir do numpy com distribuições das chaves do dicionário

## Passo 4 – Calcular estatísticas básicas
Para cada distribuição, calcule:  
- Média  
- Mediana  
- Variância  
- Desvio padrão  
Armazene em um dicionário dentro da chave `"estatisticas"`.  

```python
distribuicoes['Normal']['estatisticas'] = {}
distribuicoes['Exponencial']['estatisticas'] = {}
distribuicoes['Uniforme']['estatisticas'] = {}
```

### Normal

```python
distribuicoes['Normal']['estatisticas']['media'] = np.mean(distribuicoes['Normal']['dados'])
distribuicoes['Normal']['estatisticas']['mediana'] = np.median(distribuicoes['Normal']['dados'])
distribuicoes['Normal']['estatisticas']['variancia'] = np.var(distribuicoes['Normal']['dados'])
distribuicoes['Normal']['estatisticas']['desvio_padrao'] = np.std(distribuicoes['Normal']['dados'])
```

### Exponencial

```python
distribuicoes['Exponencial']['estatisticas']['media'] = np.mean(distribuicoes['Exponencial']['dados'])
distribuicoes['Exponencial']['estatisticas']['mediana'] = np.median(distribuicoes['Exponencial']['dados'])
distribuicoes['Exponencial']['estatisticas']['variancia'] = np.var(distribuicoes['Exponencial']['dados'])
distribuicoes['Exponencial']['estatisticas']['desvio_padrao'] = np.std(distribuicoes['Exponencial']['dados'])
```

### Uniforme

```python
distribuicoes['Uniforme']['estatisticas']['media'] = np.mean(distribuicoes['Uniforme']['dados'])
distribuicoes['Uniforme']['estatisticas']['mediana'] = np.median(distribuicoes['Uniforme']['dados'])
distribuicoes['Uniforme']['estatisticas']['variancia'] = np.var(distribuicoes['Uniforme']['dados'])
distribuicoes['Uniforme']['estatisticas']['desvio_padrao'] = np.std(distribuicoes['Uniforme']['dados'])
```

Aqui são adicionados ao dicionário de estatisticas a media, mediana, variancia e desvio padrao

## Passo 5 – Visualizar os dados
Para cada distribuição, faça um **histograma** mostrando a forma da distribuição.  

```python
plt.hist(distribuicoes['Normal']['dados'])
plt.title('Distribuição da normal gerada')
plt.hist
plt.show()
```

```python
plt.hist(distribuicoes['Exponencial']['dados'])
plt.title('Distribuição Exponencial gerada')
plt.show()
```

```python
plt.hist(distribuicoes['Uniforme']['dados'])
plt.title('Distribuição Uniforme gerada')
plt.show()
```

## Passo 6 – Organizar resultados em Pandas
Crie um `DataFrame` contendo as estatísticas de todas as distribuições.  

```python
distribuicoes['Normal']['estatisticas']
```

```python
dados_filtrados = {nome: valores['estatisticas'] for nome, valores in distribuicoes.items()}

df = pd.DataFrame.from_dict(dados_filtrados, orient='index')
df
```

## Passo 7 – Adicionar novas distribuições
- Inclua mais uma distribuição, como `"Poisson"` ou `"Binomial"`.  
- Gere dados usando `numpy.random`.  
- Calcule as estatísticas básicas e atualize o `DataFrame`.  

```python
distribuicoes['Poisson'] = {}
distribuicoes['Poisson']['dados'] = np.random.poisson(size = 1000)
distribuicoes['Poisson']['estatisticas'] = {
    'media' : np.mean(distribuicoes['Poisson']['dados']),
    'mediana' : np.median(distribuicoes['Poisson']['dados']),
    'variancia' : np.var(distribuicoes['Poisson']['dados']),
    'desvio_padrao' : np.std(distribuicoes['Poisson']['dados'])
}
```

## Passo 8 – Estatísticas adicionais
Para cada distribuição, calcule também:  
- Mínimo (`np.min`)  
- Máximo (`np.max`)  
- Quartis (`np.percentile`)  

Atualize o campo `"estatisticas"` e o `DataFrame` com essas novas colunas.  

```python
for distribuicao in distribuicoes.keys():
    dados = distribuicoes[distribuicao]['dados']

    minimo = np.min(dados)
    maximo = np.max(dados)
    quartis = np.percentile(dados, q = (25, 50, 75))
    distribuicoes[distribuicao]['estatisticas'].update({
        'minimo': minimo, 'maximo':maximo, 'quartis':quartis
    })
```

```python
dados_filtrados = {nome: valores['estatisticas'] for nome, valores in distribuicoes.items()}

df = pd.DataFrame.from_dict(dados_filtrados, orient='index')
df
```

## Passo 9 – Comparar distribuições graficamente
Faça um **boxplot** com todas as distribuições no mesmo gráfico, com título e rótulos adequados.  

```python
fig, axs = plt.subplots(1, 4, figsize=(15, 4), sharey= True)

axs[0].boxplot(distribuicoes['Normal']['dados'],)
axs[0].set_title('Normal')

axs[1].boxplot(distribuicoes['Exponencial']['dados'])
axs[1].set_title('Exponencial')

axs[2].boxplot(distribuicoes['Uniforme']['dados'])
axs[2].set_title('Uniforme')

axs[3].boxplot(distribuicoes['Poisson']['dados'])
axs[3].set_title('Poisson')

plt.tight_layout()
plt.show()
```
