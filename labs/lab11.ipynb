```python
import numpy as np

K = 100_000

def IC(dados):
    n = len(dados)
    media = np.mean(dados)
    variancia = np.var(dados, ddof=1) / n
    erro = 1.96 * np.sqrt(variancia)
    return media, variancia, (media - erro, media + erro)
```

# Exercício 1

Considere $X\sim\operatorname{Gamma}(3,2)$. Estime por Monte Carlo

$$\mu=E[\sqrt{X}].$$

Calcule a variância do estimador e construa um intervalo de confiança de 95%. Compare com o valor teórico.

**Valor teórico:** $\mu\approx 1{,}174982$.

```python
# Gamma(3,2) = soma de 3 exponenciais de taxa 2
uniformes = np.random.uniform(size=(K, 3))
gama = np.sum(-np.log(uniformes) / 2, axis=1)

valores = np.sqrt(gama)
media1, var1, ic1 = IC(valores)
print('estimativa:', media1)
print('variância do estimador:', var1)
print('IC 95%:', ic1)
print('valor teórico:', 1.174982)
```

# Exercício 2

Considere $X\sim\operatorname{Beta}(2,5)$. Estime por Monte Carlo $p=P(X>0{,}4)$.

Calcule a variância do estimador e construa um intervalo de confiança de 95%.

**Valor teórico:** $p=0{,}23328$.

```python
# Beta(2,5) usando G1/(G1+G2), com gammas geradas por uniformes
u2 = np.random.uniform(size=(K, 2))
u5 = np.random.uniform(size=(K, 5))
gama2 = np.sum(-np.log(u2), axis=1)
gama5 = np.sum(-np.log(u5), axis=1)
beta = gama2 / (gama2 + gama5)

sucessos = (beta > 0.4)
media2, var2, ic2 = IC(sucessos)
print('estimativa:', media2)
print('variância do estimador:', var2)
print('IC 95%:', ic2)
print('valor teórico:', 0.23328)
```

# Exercício 3

Considere a integral

$$I=\frac{1}{\sqrt{8\pi}}\int_{-\infty}^{\infty}(x^2+2x+1)e^{-(x-1)^2/8}\,dx.$$

Utilizando $X\sim N(1,4)$, a integral vira $I=E[X^2+2X+1]$. Gere a Normal pelo método de Box-Muller.

**Valor teórico:** $I=8$.

```python
# Box-Muller, usando só uniformes
u1, u2 = np.random.uniform(size=(2, K))
normal = np.sqrt(-2 * np.log(u1)) * np.cos(2 * np.pi * u2)
x = 1 + 2 * normal

valores = x**2 + 2*x + 1
media3, var3, ic3 = IC(valores)
print('estimativa:', media3)
print('variância do estimador:', var3)
print('IC 95%:', ic3)
print('valor teórico:', 8)
```

# Exercício 4

Considere $X\sim\operatorname{Gamma}(2,1)$ e $Y\sim\operatorname{Exp}(1)$, independentes. Estime $p=P(X>Y)$.

Calcule a variância do estimador e construa um intervalo de confiança de 95%.

**Valor teórico:** $p=0{,}75$.

```python
u = np.random.uniform(size=(K, 3))
x = np.sum(-np.log(u[:, :2]), axis=1)
y = -np.log(u[:, 2])

sucessos = (x > y)
media4, var4, ic4 = IC(sucessos)
print('estimativa:', media4)
print('variância do estimador:', var4)
print('IC 95%:', ic4)
print('valor teórico:', 0.75)
```

# Exercício 5

Considere $I=\int_0^1 x^2(1-x)^3e^x\,dx$.

Usando $X\sim\operatorname{Beta}(3,4)$, temos $I=E[e^X]/60$.

**Valor teórico:** $I\approx 0{,}02598262$.

```python
u3 = np.random.uniform(size=(K, 3))
u4 = np.random.uniform(size=(K, 4))
g3 = np.sum(-np.log(u3), axis=1)
g4 = np.sum(-np.log(u4), axis=1)
x = g3 / (g3 + g4)

valores = np.exp(x) / 60
media5, var5, ic5 = IC(valores)
print('estimativa:', media5)
print('variância do estimador:', var5)
print('IC 95%:', ic5)
print('valor teórico:', 0.02598262)
```

# Exercício 6

Considere $X,Y\overset{\mathrm{iid}}{\sim}N(0,1)$. Estime $p=P(X^2+Y^2>5)$.

Gere as duas Normais pelo método de Box-Muller e compare com o valor da qui-quadrado.

**Valor teórico:** $p=e^{-5/2}\approx 0{,}082085$.

```python
u = np.random.uniform(size=(4, K))
x = np.sqrt(-2*np.log(u[0])) * np.cos(2*np.pi*u[1])
y = np.sqrt(-2*np.log(u[2])) * np.cos(2*np.pi*u[3])

sucessos = (x**2 + y**2 > 5)
media6, var6, ic6 = IC(sucessos)
print('estimativa:', media6)
print('variância do estimador:', var6)
print('IC 95%:', ic6)
print('valor teórico:', np.exp(-5/2))
```

# Exercício 7

Considere $X_1,\ldots,X_5$ independentes, com $X_i\sim\operatorname{Exp}(2)$. Defina $M=\max(X_1,\ldots,X_5)$.

Estime $E[M]$ por Monte Carlo e faça um intervalo de confiança de 95%.

**Valor teórico:** $E[M]\approx 1{,}141667$.

```python
u = np.random.uniform(size=(K, 5))
amostras = -np.log(u) / 2
maior = np.max(amostras, axis=1)

media7, var7, ic7 = IC(maior)
print('estimativa:', media7)
print('variância do estimador:', var7)
print('IC 95%:', ic7)
print('valor teórico:', 1.141667)
```

# Exercício 8

Considere $X=e^Z$, com $Z\sim N(0,1)$. Estime $\mu=E[X]$ e $p=P(X>10)$.

**Valores teóricos:** $\mu=e^{1/2}\approx 1{,}648721$ e $p\approx 0{,}0106511$.

```python
u1, u2 = np.random.uniform(size=(2, K))
z = np.sqrt(-2*np.log(u1)) * np.cos(2*np.pi*u2)
x = np.exp(z)

media8, var8, ic8 = IC(x)
print('estimativa de E[X]:', media8)
print('variância do estimador da média:', var8)
print('IC 95% da média:', ic8)

sucessos = (x > 10)
prob8, varp8, icp8 = IC(sucessos)
print('estimativa de P(X > 10):', prob8)
print('variância do estimador da probabilidade:', varp8)
print('IC 95% da probabilidade:', icp8)
print('valores teóricos:', np.exp(0.5), 0.0106511)
```
