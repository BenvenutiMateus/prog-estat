### Exercício 1 — Estimação de integrais

Considere a integral

$$
I=\int_0^1 e^{-x^2}\,dx.
$$

Utilizando $X\sim U(0,1)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

Isso é uma $U(0,1)$ e $E[e^{-x^2}]$

```python
n = 10 ** 4
unis = np.random.uniform(size=n)
amostra = np.exp(-unis**2)
integral = np.mean(amostra)
```

```python
integral
```

```python
s_quadrado = np.var(amostra, ddof=1)

s = np.sqrt(s_quadrado) 

z = 1.96
margem_erro = z * s / np.sqrt(n)
ic_integral = IC(amostra)

print("Intervalo de Confiança:", ic_integral)
```

### Exercício 2 — Estimação de probabilidades

Considere $X\sim N(0,1)$ e a probabilidade

$$
p=P(X>1).
$$

Estime $p$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

```python
c = 2/np.sqrt(2*np.pi)*np.exp(1/2)
c
```

```python
amostra_normal = []
sucessos = 0

while sucessos < n:
    y = exp(M = 1)[0]
    u = np.random.uniform()
    if u < densidade_normal(y)/(densidade_exponencial(y)*c):
        m = np.random.uniform()
        amostra_normal.append(y* (-1)**(m < .5))
        sucessos += 1
amostra_normal = np.array(amostra_normal)
```

```python
maior_1 = amostra_normal > 1
p_maior_1 = np.mean(maior_1)
p_maior_1
```

```python
ic_maior_1 = IC(maior_1)
ic_maior_1
```
