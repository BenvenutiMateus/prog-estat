```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
n = 100 # Tamanho do agrupamento para o TLC
n_sim = 500000# Número de simulações/amostras
p = 0.5

# 1. Gerar 'n_sim' amostras, cada uma sendo a média de 'n' variáveis uniformes U(0,1)
amostragem = np.random.binomial(n, p, (n_sim, n))
X_barra = np.mean(amostragem, axis=1)

# 2. Parâmetros teóricos da distribuição Uniforme(0,1)
mu = p
sigma = np.sqrt(n*p*(1-p))

# 3. Normalização correta usando os parâmetros TEÓRICOS
norm = np.sqrt(n) * (X_barra - mu) / sigma

# 4. Plotar o histograma
plt.hist(norm, bins = 1000, density=True,alpha=0.6, color='g')

# Curva teórica Normal(0,1) para comparação
x = np.linspace(-4, 4, 100)
plt.plot(x, 1/np.sqrt(2*np.pi) * np.exp(-x**2 / 2), 'r--', lw=2)

plt.title('Teorema do Limite Central')
plt.show()
```
