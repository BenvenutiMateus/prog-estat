````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

````python
distribuicoes = {
    "Exponencial": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
````

````python
normal = np.random.normal(**distribuicoes["Normal"]['params'], size=1000)
exp = np.random.exponential(**distribuicoes["Exponencial"]['params'], size = 1000)
uni = np.random.uniform(**distribuicoes["Uniforme"]['params'], size= 1000)
distribuicoes["Normal"]["dados"] = normal
distribuicoes["Exponencial"]["dados"] = exp
distribuicoes["Uniforme"]["dados"] = uni
````

### Normal

````python
media = np.mean(distribuicoes['Normal']['dados'])
mediana = np.median(distribuicoes['Normal']['dados'])
var = np.var(distribuicoes['Normal']['dados'])
desvio_padrao = np.std(distribuicoes['Normal']['dados'])

distribuicoes['Normal']['estatisticas'] = {'media' : media, 'mediana' : mediana, 'variancia' :var, 'desvio_padrao' : desvio_padrao}
````

## Exponencial

````python
media = np.mean(distribuicoes['Exponencial']['dados'])
mediana = np.median(distribuicoes['Exponencial']['dados'])
var = np.var(distribuicoes['Exponencial']['dados'])
desvio_padrao = np.std(distribuicoes['Exponencial']['dados'])

distribuicoes['Exponencial']['estatisticas'] = {'media' : media, 'mediana' : mediana, 'variancia' :var, 'desvio_padrao' : desvio_padrao}
````

````python
mediana = np.median(distribuicoes['Exponencial']['dados'])
var = np.var(distribuicoes['Exponencial']['dados'])
desvio_padrao = np.std(distribuicoes['Exponencial']['dados'])

distribuicoes['Exponencial']['estatisticas'] = {'media' : media, 'mediana' : mediana, 'variancia' :var, 'desvio_padrao' : desvio_padrao}
````

## Uniforme

````python
media = np.mean(distribuicoes['Uniforme']['dados'])
mediana = np.median(distribuicoes['Uniforme']['dados'])
var = np.var(distribuicoes['Uniforme']['dados'])
desvio_padrao = np.std(distribuicoes['Uniforme']['dados'])

distribuicoes['Uniforme']['estatisticas'] = {'media' : media, 'mediana' : mediana, 'variancia' :var, 'desvio_padrao' : desvio_padrao}
````
